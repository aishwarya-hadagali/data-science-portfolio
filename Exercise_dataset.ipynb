```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = sns.load_dataset('exercise')
df.head()
df.isnull().sum()
df.duplicated().sum()
df.drop_duplicates(inplace=True)
df.duplicated().sum()
df.info()
df.describe()

df.drop('Unnamed: 0', axis=1, inplace=True)
df.drop('id',axis=1, inplace=True)
df.isnull().sum()
df.shape

df.duplicated().sum()
df.isnull().sum()
df.describe(include='category')

#univariate plot
fig,axes = plt.subplots(2,2,figsize=(15,10))
sns.histplot(x='diet',data=df,ax=axes[0,0])
sns.countplot(x='time',data=df,ax=axes[0,1])
sns.countplot(x='kind',data=df,ax=axes[1,0])
sns.countplot(x='pulse',data=df,ax=axes[1,1])
plt.show()

axes[0,0].set_title('Diet')
axes[0,1].set_title('Time')
axes[1,0].set_title('Kind')
axes[1,1].set_title('Pulse')
plt.show()

#bivariate plots
fig,axes = plt.subplots(1,2,figsize=(12,10))
sns.boxplot(x='kind',y='pulse',data=df,ax=axes[0])
sns.boxplot(x='diet',y='pulse',data=df,ax=axes[1])
plt.show()

axes[0].set_title('Kind')
axes[1].set_title('Diet')
plt.show()
```
