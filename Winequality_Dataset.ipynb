```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/WineQT.csv', sep=',')
df.head()
df.shape
df.info()
df.isnull().sum()
df.duplicated().sum()
df.dropna(subset=['Id'],inplace=True)

df.describe()

#univariate
fig,axes = plt.subplots(2,2,figsize=(15,10))
sns.histplot(data=df,x='quality',ax=axes[0,0])
sns.histplot(data=df,x='alcohol',ax=axes[0,1])
sns.histplot(data=df, x='volatile acidity',ax=axes[1,0])
sns.histplot(data=df,x='residual sugar',ax=axes[1,1])

axes[0,0].set_title('quality')
axes[0,1].set_title('alcohol')
axes[1,0].set_title('volatile acidity')
axes[1,1].set_title('residual sugar')

plt.tight_layout()
plt.show()

#bivariate
fig,axes = plt.subplots(1,2,figsize=(15,5))
sns.boxplot(data=df,x='quality',y='alcohol',ax=axes[0])
sns.boxplot(data=df,x='quality',y='volatile acidity',ax=axes[1])

axes[0].set_title('alcohol vs quality')
axes[1].set_title('volatile acidity vs quality')

plt.tight_layout()
plt.show()
```

🍷 EDA on Wine Quality Dataset | Aishwarya Hadagali

📌 Dataset: 1,143 red wines, 12 chemical properties

Key Insights:
1. 🎯 Most wines are average — quality scores 5 and 6
   dominate. Very few wines rated 3 or 8.
   Great wine is rare!

2. 🍾 Alcohol boosts quality — higher quality wines
   consistently have more alcohol (~12% for quality 8
   vs ~10% for quality 3). More alcohol = better taste!

3. 🧪 Acidity kills quality — volatile acidity drops
   sharply as quality increases. Too much acidity
   makes wine taste like vinegar!

4. 🍬 Sugar is low for most wines — residual sugar
   is heavily right-skewed. Most red wines are dry
   with very little sugar (2-3g).

5. 💡 Key finding — alcohol and volatile acidity are
   the two strongest indicators of wine quality.
   These could be powerful features for an ML model!

🛠 Tools used: Python, Pandas, Seaborn, Matplotlib
