```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from plotly.subplots import make_subplots
import plotly.graph_objects as go
import plotly.express as px


df = pd.read_csv('/content/diabetes.csv')
df.head()
df.shape
df.info()
df.duplicated().sum()

df.describe()

#univariat

# Create 2x2 subplot grid
fig = make_subplots(rows=2, cols=2,
                    subplot_titles=('Glucose Distribution',
                                   'BMI Distribution',
                                   'Age Distribution',
                                   'Outcome Count'))

# Plot 1 — Glucose histogram (row 1, col 1)
fig.add_trace(go.Histogram(x=df['Glucose'], name='Glucose'), row=1, col=1)

# Plot 2 — BMI histogram (row 1, col 2)
fig.add_trace(go.Histogram(x=df['BMI'], name='BMI'), row=1, col=2)

# Plot 3 — Age histogram (row 2, col 1)
fig.add_trace(go.Histogram(x=df['Age'], name='Age'), row=2, col=1)

# Plot 4 — Outcome countplot (row 2, col 2)
outcome_counts = df['Outcome'].value_counts()
fig.add_trace(go.Bar(x=['No Diabetes', 'Diabetes'],
                     y=outcome_counts.values,
                     name='Outcome'), row=2, col=2)

# Update layout
fig.update_layout(title='Diabetes Dataset — Univariate Analysis',
                  height=600, showlegend=False)

fig.show()

#bivariate

fig = make_subplots(rows=2, cols=2,
                    subplot_titles=('Glucose vs Outcome',
                                   'BMI vs Outcome',
                                   'Age vs Outcome',
                                   'Insulin vs Outcome'))

fig.add_trace(go.Box(x=df['Outcome'], y=df['Glucose']), row=1, col=1)
fig.add_trace(go.Box(x=df['Outcome'], y=df['BMI']), row=1, col=2)
fig.add_trace(go.Box(x=df['Outcome'], y=df['Age']), row=2, col=1)
fig.add_trace(go.Box(x=df['Outcome'], y=df['Insulin']), row=2, col=2)

fig.update_layout(title='Diabetes Dataset — Bivariate Analysis',
                  height=600, showlegend=False)
fig.show()
```
